## Gap-Check Findings (Pre-Cleaning Diagnostics)

Before starting the cleaning phase, we ran two additional checks to validate assumptions made in the initial summary.

**1. Cancellations vs. negative Quantity**
- 10,206 invoices start with "C" (cancellations); 12,326 rows have negative Quantity — the gap is expected since one invoice can have multiple line items.
- 1 C-invoice has a positive quantity (negligible, likely a fee line inside a cancelled order).
- **2,121 rows have negative Quantity but are NOT C-invoices.** This means "negative Quantity = cancellation" was not fully accurate — these rows are a separate category, likely stock write-offs or inventory corrections, and need their own handling in the cleaning phase rather than being grouped with customer cancellations.

**2. Non-product StockCodes**
- An initial scan for short/all-letter StockCodes flagged 17 unusual codes. Row counts and Descriptions were checked for each to confirm which are genuine operational entries vs. real products with irregular codes.
- **12 codes (2,856 rows, ~0.54% of the dataset) are operational, not product sales:** `POST`, `DOT`, `M`/`m`, `C2`, `C3`, `D`, `ADJUST`, `S`, `AMAZONFEE`, `B`, `GIFT`. These cover postage, carriage, discounts, manual entries, samples, fees, and bad-debt adjustments, and should be excluded from product/revenue analysis.
- **5 codes initially flagged as "unusual" are actually real products:** `PADS`, `DCGSSGIRL`, `DCGSSBOY`, `DCGSLBOY`, `DCGSLGIRL` — a party bag product line (boy/girl, size S/L variants). These should be kept.
- Takeaway: pattern-based filters (short codes, all-letter codes) are a useful starting point but need manual review before any rows are dropped — several legitimate products would have been lost otherwise.

**Conclusion:** The cleaning phase needs to separately account for (a) customer cancellations, (b) non-"C" negative-quantity write-offs, and (c) non-product StockCodes, rather than treating "negative Quantity" and "unusual StockCode" as single catch-all categories.

## Cleaning Decisions (Pre-Cleaning Plan)

Before writing any cleaning code, the following decisions were made based on the diagnostics above and the intended analysis direction.

**Analysis direction:** Primary focus on sales trends and product performance over time. A secondary, lighter-weight customer view (e.g. top customers, repeat vs. one-time buyers) will be added afterward using only the subset of data with a valid Customer ID. This direction was chosen because Quantity, Price, InvoiceDate, and StockCode are 100% complete, while Customer ID is missing on ~20% of rows — building the core analysis on the more complete fields gives a stronger, more defensible result.

| Issue | Decision | Reasoning |
|---|---|---|
| Duplicate rows (6,865) | Drop | Exact duplicates inflate every downstream count/aggregate with no analytical value |
| Non-product StockCodes (12 codes, 2,856 rows: `POST`, `DOT`, `M`/`m`, `C2`, `C3`, `D`, `ADJUST`, `S`, `AMAZONFEE`, `B`, `GIFT`) | Drop from product/revenue analysis | These represent postage, fees, discounts, and manual adjustments — not real product sales |
| Real products with irregular codes (`PADS`, `DCGSSGIRL`, `DCGSSBOY`, `DCGSLBOY`, `DCGSLGIRL`) | Keep | Confirmed genuine products via Description; excluding them would lose real sales data |
| Cancellations (10,205 "C"-prefixed invoices) | Separate into their own subset | Represent customer-cancelled orders; excluded from gross sales but may be reported separately (e.g. cancellation rate by product) |
| Non-"C" negative-quantity rows (2,121) | Separate into their own subset | Likely stock write-offs/corrections, not customer cancellations; treated as a distinct category rather than merged with cancellations |
| Negative/zero Price on remaining product rows | Investigate after non-product codes are removed, then drop confirmed errors | Most extreme Price values were traced to `DOT`/`M` rows; remaining cases need a fresh check |
| Missing Customer ID (107,927 rows, 20.5%) | Keep rows; exclude only from the customer-level sub-analysis | Dropping these would remove valid Quantity/Price/Date data needed for the primary (trend/product) analysis, which doesn't depend on Customer ID |
| Missing Description (2,928 rows) | Low priority — drop or leave as-is | Small volume (0.56%), doesn't block quantitative analysis |
| Data types | Convert `Customer ID` to string/int (it's an identifier, not a quantity); confirm `StockCode` as string | Prevents accidental numeric operations on ID fields |

**Note:** These decisions are scoped to the current analysis direction. If the project later expands to include deeper customer segmentation (e.g. RFM), the Customer ID decision should be revisited.

```python
df['Revenue'] = df['Quantity'] * df['Price']
# total money earned per row: units sold x price per unit
# this becomes the base for every other revenue-related feature (product, invoice, customer level)

df[['Quantity', 'Price', 'Revenue']].head()
```

```python
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['MonthName'] = df['InvoiceDate'].dt.month_name()
df['YearMonth'] = df['InvoiceDate'].dt.to_period('M').astype(str)
# YearMonth (e.g. "2009-2010") groups cleanly for a monthly trend line without year/month mismatches

df['DayOfWeek'] = df['InvoiceDate'].dt.day_name()
df['Hour'] = df['InvoiceDate'].dt.hour

df[['InvoiceDate', 'Year', 'Month', 'MonthName', 'YearMonth', 'DayOfWeek', 'Hour']].head()
```

```python
product_agg = df.groupby('StockCode').agg(
    Description=('Description', 'first'),
    TotalQuantitySold=('Quantity', 'sum'),
    TotalRevenue=('Revenue', 'sum'),
    NumInvoices=('Invoice', 'nunique'),
    NumUniqueCustomers=('Customer ID', 'nunique')
).reset_index()
# one row per product, with quantity/revenue/reach/reach-among-customers summarized
# NumUniqueCustomers only counts customers with a valid ID (nunique ignores NaN automatically)

product_agg = product_agg.sort_values('TotalRevenue', ascending=False)
product_agg.head(10)
```

```python
invoice_agg = df.groupby('Invoice').agg(
    InvoiceDate=('InvoiceDate', 'first'),
    ItemsPerInvoice=('StockCode', 'count'),
    InvoiceTotal=('Revenue', 'sum')
).reset_index()
# one row per invoice/order
# ItemsPerInvoice = number of line items in that order (not units - a proxy for basket size)
# InvoiceTotal = total revenue for that order

invoice_agg.describe()
```

```python
customer_df = df[df['Customer ID'].notna()]
# subset with valid Customer ID only, as agreed - ~80% of transactions

customer_agg = customer_df.groupby('Customer ID').agg(
    TotalSpend=('Revenue', 'sum'),
    TotalOrders=('Invoice', 'nunique')
).reset_index()
# one row per customer

customer_agg['CustomerType'] = customer_agg['TotalOrders'].apply(
    lambda x: 'Repeat' if x > 1 else 'One-time'
)
# simple repeat vs one-time flag, no RFM needed

customer_agg.head()
```

```python
# repeat vs one-time split, in counts
customer_agg['CustomerType'].value_counts()
```

```python
# revenue share: repeat buyers vs one-time buyers
customer_agg.groupby('CustomerType')['TotalSpend'].sum()
```

## Sales Trends & Product Performance Summary

Analysis was conducted on the cleaned Online Retail II dataset (502,512 rows), using the engineered features from the previous section. Six analyses were performed, each paired with a supporting visualization.

### Sales Trends

1. **Monthly revenue trend** — Revenue builds steadily from August through November 2010, peaking in November — consistent with pre-Christmas ordering for a gift/homeware wholesaler. Dec 2009 (partial month, data starts Dec 1) and Dec 2010 (data ends Dec 9) were excluded from this view to avoid a misleading artificial drop; December revenue likely would have continued the upward trend had full data been available.

2. **Day-of-week pattern** — Revenue is fairly steady Monday–Thursday (peaking Thursday), dips on Friday, and Sunday is notably strong for a wholesale platform. Saturday shows almost no invoice activity (400 of ~502,512 rows) — this reflects the platform not processing orders that day, not a customer behavior pattern, and should not be read as a "low sales day."

3. **Hour-of-day pattern** — A clean business-hours curve: activity begins around 7am, peaks at noon, and tapers off by 8pm. This confirms B2B ordering behavior (orders placed during business hours) rather than consumer shopping patterns.

### Product Performance

## 3. SMART Goal Formulation

#### Goal DQ1 — Zero duplicates in future analysis-ready datasets

- **Specific:** Apply exact-row duplicate validation before any feature engineering or aggregate analysis in every future dataset refresh.

- **Achievable:** This is realistic because the notebook has already demonstrated the required `drop_duplicates()` operation.
- **Relevant:** Duplicate rows artificially increase transaction counts, quantities, and revenue aggregates, so preventing them protects the accuracy of every later finding.
- **Time-bound:** Apply from the **next dataset refresh onward** and verify before every future analysis run.

#### Goal DQ2 — Reduce missing Customer IDs

- **Specific:** Improve Customer ID capture for transactions that would otherwise be excluded from customer-level analysis.
- **Measurable:** Use the missing-value count in `Customer ID`. Reduce missing IDs from **102,911 rows** to **no more than 90,000 rows**, assuming a comparable dataset size.
